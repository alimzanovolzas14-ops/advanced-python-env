```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Name": ["Danial", "Dias", "Olzhas", "Madina", "Nursultan"],
    "Score": [85, 78, 92, 88, 75]
}

df = pd.DataFrame(data)
```

```python
plt.plot(df["Name"], df["Score"], marker='o')
plt.xlabel("Students")
plt.ylabel("Scores")
plt.title("Student Scores Line Plot")
plt.show()
```

```python
plt.bar(df["Name"], df["Score"])
plt.xlabel("Students")
plt.ylabel("Scores")
plt.title("Student Scores Bar Chart")
plt.show()
```
